```python
import numpy as np

#  A = [a b] = [1 2]
#      [c d]   [5 4]
#  det(A) = ad - bc

A = np.array([[1,2],[5,4]])
print(np.asmatrix(A))
A, np.linalg.det(A)
```

```python
from scipy import sparse as sp

# 0.70710678 = 1/√2 (what?)

evls, evts = sp.linalg.eigs(A)

for i in range(2): 
    λ = evls[i]
    v = evts[:, i]
    print(f"evl = {λ}, evt = {v}")
    I0 = np.identity(2) * λ
    A0 = A - I0
    assert (A0 @ v == 0).all()


```
