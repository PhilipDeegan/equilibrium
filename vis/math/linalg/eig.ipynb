```python
# https://math.libretexts.org/Bookshelves/Linear_Algebra/A_First_Course_in_Linear_Algebra_(Kuttler)/07%3A_Spectral_Theory/7.01%3A_Eigenvalues_and_Eigenvectors_of_a_Matrix
# https://www.geeksforgeeks.org/eigen-values/#eigenvectors-of-a-square-matrix

import numpy as np

#  A = [a b] = [1 2]
#      [c d]   [5 4]
#  det(A) = ad - bc

A = np.array([[1,2],[5,4]])
A, np.linalg.det(A)
```

```python
# eigenvalues and eigenvectors of A

#  λ = eigenvalue
#  I = identity matrix (ones on the diagonal)
#  IF 
#    |A - λI| = 0
#
#  So
#    [1 2] - λ[1 0] = 0
#    [5 4]    [0 1]
#
#    [1-λ   2] = 0
#    [5   4-λ]
#
#    matrix determinant expansion...
#
#         (1-λ)(4-λ) - 10 = 0
#    4 - λ - 4λ + λ² - 10 = 0
#             λ² - 5λ - 6 = 0
#
#    quadratic formula...
#
#     x = (5 + root(25 + 24)) / 2 OR x = (5 - root(25 + 24)) / 2
#                 x = (5 + 7) / 2 OR x = (5 - 7) / 2
#                           x = 6 OR x = -1
#
#    λ = 6 or λ = -1

evls = [-1, 6]
```

```python
# eigenvectors of A with ascertained eigenvalues

#  λ = eigenvalue
#  I = identity matrix (ones on the diagonal)
#  v = eigenvector = [a]
#                    [b]
#
#  forumla..  
#
#    (A-λI)v = 0
#
#  So for λ = 6
#
#    ([1 2] - 6[1 0]).[a] = 0
#    ([5 4]    [0 1]) [b]
#
#    [-5  2].[a] = 0
#    [ 5 -2] [b]
#  
#    [-5a + 2b] = 0
#    [ 5a - 2b]
#
#    -5a*-2b - (2b*5a) = 0
#
#    a=2, b=5
#
#    v = [2]
#        [5]
#
#     for λ = -1
#
#    v = [ 1]
#        [-1]
#

evts = [[1,-1], [2,5]]
```

```python

for λ, v in zip(evls, evts):
    print(f"evl = {λ}, evt = {v}")
    A0 = A - (np.identity(2) * λ)
    np.testing.assert_allclose(A0 @ v, 0, rtol=0, atol=1e-16)
    assert (A0 @ v == 0).all() # binary ==
    
```
